```python
import numpy as np

# Read the CSV
df = pd.read_csv("output/ai_articles_updated.csv")

# Extract company names from title/summary if company column is empty
def extract_company(row):
    """Extract company from company column or title/summary"""
    # If company column exists and is not empty
    if 'company' in row.index and pd.notna(row.get('company')) and row['company'] != '':
        return row['company']
    
    # Otherwise extract from title/summary
    text = f"{row.get('title', '')} {row.get('summary', '')}".lower()
    
    companies = ['Octaplus,BT, VMO2, Vodafone, Sky, TalkTalk, Glide, Zen, Utility Warehouse, Cuckoo, YouFibre / Brsk, Openreach, CityFibre, Netomnia, nexfibre, Hyperoptic, Community Fibre, FullFibre, Gigaclear, APFN, ITS']
    
    for company in companies:
        if company.lower() in text:
            return company
    
    return 'Other/Unknown'

# Extract companies
df['company'] = df.apply(extract_company, axis=1)

# Define the 6 use case categories
use_cases = [
    'network_planning_deployment',
    'network_optimization_performance',
    'predictive_maintenance_analytics',
    'customer_support_experience',
    'security_fraud_threat',
    'internal_productivity_workforce'
]

# Create the pivot table (Company × Use Case matrix)
matrix = pd.crosstab(
    df['company'],
    df['primary_use_case'],
    margins=False
)
```

```python


# Reorder columns to match the 6 categories
matrix = matrix[use_cases]

# Sort by total count (descending)
matrix['Total'] = matrix.sum(axis=1)
matrix = matrix.sort_values('Total', ascending=False)

# Remove Total column for display (or keep it)
display_matrix = matrix.drop('Total', axis=1) if 'Total' in matrix.columns else matrix

# Print the matrix
print("=" * 100)
print("COMPANY × USE CASE MATRIX")
print("=" * 100)
print(f"\nTotal articles analyzed: {len(df)}")
print(f"Companies: {len(matrix)}")
print("\n" + "=" * 100)

# Format for better readability
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', 20)

print("\nMatrix (Count of articles per company-use case combination):")
print(display_matrix)

# Also show with shortened column names
short_names = {
    'network_planning_deployment': 'Network Planning',
    'network_optimization_performance': 'Network Opt',
    'predictive_maintenance_analytics': 'Predictive Maint',
    'customer_support_experience': 'Customer Support',
    'security_fraud_threat': 'Security/Fraud',
    'internal_productivity_workforce': 'Internal Ops'
}

matrix_short = display_matrix.rename(columns=short_names)
print("\n" + "=" * 100)
print("Matrix with Shortened Names:")
print(matrix_short)

# Add totals row
matrix_with_totals = display_matrix.copy()
matrix_with_totals.loc['TOTAL'] = matrix_with_totals.sum()
print("\n" + "=" * 100)
print("Matrix with Totals:")
print(matrix_with_totals.rename(columns=short_names))

# Save to CSV
matrix_with_totals.rename(columns=short_names).to_csv("company_use_case_matrix.csv")
print("\n✅ Saved to: company_use_case_matrix.csv")
```

```python

matrix = pd.read_csv("company_use_case_matrix.csv", index_col=0)

# Convert to binary
binary_matrix = (matrix > 0).astype(int)

# Handle TOTAL row if present
if 'TOTAL' in binary_matrix.index:
    # Remove old TOTAL, recalculate
    binary_matrix = binary_matrix.drop('TOTAL')
    binary_matrix.loc['TOTAL'] = binary_matrix.sum()

# Display
print("BINARY COMPANY × USE CASE MATRIX")
print("=" * 80)
print("(1 = Company has this use case, 0 = Company doesn't have)")
print()
print(binary_matrix)

# Save
binary_matrix.to_csv("company_use_case_matrix_binary.csv")
print("\n✅ Saved to: company_use_case_matrix_binary.csv")
```

```python


# Read the binary matrix
binary_matrix = pd.read_csv("company_use_case_matrix_binary.csv", index_col=0)

# Calculate coverage (sum of each row = number of categories covered)
binary_matrix['Coverage'] = binary_matrix.sum(axis=1)

# Sort by coverage (descending) - companies with most categories first
binary_matrix = binary_matrix.sort_values('Coverage', ascending=False)

# Remove Coverage column for display (or keep it)
display_matrix = binary_matrix.drop('Coverage', axis=1) if 'Coverage' in binary_matrix.columns else binary_matrix

print("BINARY MATRIX SORTED BY CATEGORY COVERAGE")
print("=" * 100)
print("(Sorted by number of categories covered - highest first)")
print()

# Show coverage stats
coverage_stats = binary_matrix['Coverage'].sort_values(ascending=False)
print("Category Coverage by Company:")
for company, coverage in coverage_stats.items():
    if company != 'TOTAL':  # Skip TOTAL row if present
        print(f"  {company}: {int(coverage)}/6 categories ({coverage/6*100:.1f}%)")

print("\n" + "=" * 100)
print("\nMatrix (sorted by coverage):")
print(display_matrix)

# Save sorted matrix
display_matrix.to_csv("company_use_case_matrix_binary_sorted.csv")
print("\n✅ Saved sorted binary matrix to: company_use_case_matrix_binary_sorted.csv")
```
